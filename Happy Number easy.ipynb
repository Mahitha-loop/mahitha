```python
def isHappy(n):
    seen = set()

    while n != 1 and n not in seen:
        seen.add(n)
        total = 0

        while n > 0:
            digit = n % 10
            total += digit ** 2
            n //= 10

        n = total

    return n == 1


# Example 1
n = 19
print(isHappy(n))  # Output: True

# Example 2
n = 2
print(isHappy(n))  # Output: False
```
