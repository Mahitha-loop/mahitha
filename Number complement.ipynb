```python
def findComplement(num):
    binary = bin(num)[2:]
    complement = ""

    for bit in binary:
        if bit == '0':
            complement += '1'
        else:
            complement += '0'

    return int(complement, 2)


# Example 1
num = 5
print(findComplement(num))  # Output: 2

# Example 2
num = 1
print(findComplement(num))  # Output: 0
```
