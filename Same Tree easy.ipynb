```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def isSameTree(self, p, q):
        if p is None and q is None:
            return True

        if p is None or q is None:
            return False

        if p.val != q.val:
            return False

        return (self.isSameTree(p.left, q.left) and
                self.isSameTree(p.right, q.right))


# Example 1: Both trees are identical
p = TreeNode(1, TreeNode(2), TreeNode(3))
q = TreeNode(1, TreeNode(2), TreeNode(3))
print(Solution().isSameTree(p, q))  # Output: True

# Example 2: Different tree structures
p = TreeNode(1, TreeNode(2), None)
q = TreeNode(1, None, TreeNode(2))
print(Solution().isSameTree(p, q))  # Output: False

# Example 3: Different node positions
p = TreeNode(1, TreeNode(2), TreeNode(1))
q = TreeNode(1, TreeNode(1), TreeNode(2))
print(Solution().isSameTree(p, q))  # Output: False
```
