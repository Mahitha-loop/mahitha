```python
def isUgly(n):
    if n <= 0:
        return False

    for factor in [2, 3, 5]:
        while n % factor == 0:
            n //= factor

    return n == 1


# Example 1
n = 6
print(isUgly(n))  # Output: True

# Example 2
n = 1
print(isUgly(n))  # Output: True

# Example 3
n = 14
print(isUgly(n))  # Output: False
```
