```python
def findEmployees(employee):
    employees = {e["id"]: e for e in employee}
    result = []

    for e in employee:
        manager_id = e["managerId"]

        if manager_id is not None and manager_id in employees:
            manager = employees[manager_id]

            if e["salary"] > manager["salary"]:
                result.append(e["name"])

    return result


employee = [
    {"id": 1, "name": "Joe", "salary": 70000, "managerId": 3},
    {"id": 2, "name": "Henry", "salary": 80000, "managerId": 4},
    {"id": 3, "name": "Sam", "salary": 60000, "managerId": None},
    {"id": 4, "name": "Max", "salary": 90000, "managerId": None}
]

print(findEmployees(employee))
```
